```python
import pickle

with open('legal_chunks_docling_20251224_154949.pkl', 'rb') as file:
    data = pickle.load(file)
    
# First, let's see the structure of the data
print("Keys in data:", data.keys() if isinstance(data, dict) else "Data is not a dict")
print("\nType of data:", type(data))
```

```python
# Function to search for specific articles in chunks
def search_articles(chunks, article_numbers):
    """
    Search for specific articles in the chunks data.
    article_numbers: list of article numbers to search for (e.g., ['Article 5', '5', 'Art. 5'])
    """
    found_articles = []
    
    # Assuming data is a dict with 'chunks_summary' key containing chunks
    chunks = data.get('chunks_summary', []) if isinstance(data, dict) else data
    
    for i, chunk in enumerate(chunks):
        chunk_text = str(chunk).lower()
        
        for article in article_numbers:
            if article.lower() in chunk_text:
                found_articles.append({
                    'chunk_index': i,
                    'article': article,
                    'chunk_preview': str(chunk)[:200] + '...' if len(str(chunk)) > 200 else str(chunk)
                })
    
    return found_articles

# Example: Search for specific articles
articles_to_find = ['Article 32', 'Article 34']  # Modify this list
results = search_articles(data, articles_to_find)

print(f"Found {len(results)} matches:")
for result in results:
    print(f"\nChunk {result['chunk_index']} contains '{result['article']}':")
    print(result['chunk_preview'])
```
